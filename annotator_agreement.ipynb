```python
import pandas as pd
import sklearn
```

```python
data = pd.read_excel('validation_set_articulation.xlsx')
```

```python
data
```

```python
labels_1 = []
labels_2 = []

for i in range(len(data)):
  labels_1.append(data['Vishal\'s Codes'][i])
  labels_2.append(data['Jack\'s Codes'][i])
```

```python
len(labels_1)
```

```python
len(labels_2)
```

```python
from sklearn.metrics import cohen_kappa_score

cohen_kappa_score(labels_1, labels_2)
```
